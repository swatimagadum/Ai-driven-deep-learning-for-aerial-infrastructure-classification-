## Step 2: Generate Sample Dataset Directories & Synthetic Images

```python
import os
import cv2
import numpy as np

CLASSES = [
    'Airforce Base', 'Army Base', 'Bridge', 'Capital City', 'Commercial Airport',
    'National Highway', 'Navy Base', 'Oil Reservoir Tank', 'Port', 'Railway Station'
]

DATASET_DIR = 'dataset'
os.makedirs(DATASET_DIR, exist_ok=True)

for cls in CLASSES:
    cls_dir = os.path.join(DATASET_DIR, cls)
    os.makedirs(cls_dir, exist_ok=True)
    for i in range(15):
        img = np.zeros((224, 224, 3), dtype=np.uint8)
        np.random.seed(sum(ord(c) for c in cls) + i)
        color = np.random.randint(50, 200, size=(3,)).tolist()
        img[:] = color
        cv2.putText(img, cls[:8], (10, 115), cv2.FONT_HERSHEY_SIMPLEX, 0.6, (255, 255, 255), 2)
        cv2.imwrite(os.path.join(cls_dir, f'sample_{i+1}.jpg'), img)

print('[SUCCESS] Dataset structure and sample images generated!')
```

## Step 3: Data Preprocessing & Augmentation

```python
from tensorflow.keras.preprocessing.image import ImageDataGenerator

datagen = ImageDataGenerator(
    rescale=1./255,
    rotation_range=30,
    width_shift_range=0.1,
    height_shift_range=0.1,
    shear_range=0.2,
    zoom_range=0.2,
    horizontal_flip=True,
    vertical_flip=True,
    validation_split=0.2
)

train_gen = datagen.flow_from_directory('dataset', target_size=(224, 224), batch_size=8, class_mode='categorical', subset='training')
val_gen = datagen.flow_from_directory('dataset', target_size=(224, 224), batch_size=8, class_mode='categorical', subset='validation', shuffle=False)
```

## Step 4: Model Training (MobileNetV2, VGG16, DenseNet121)

```python
import tensorflow as tf
from tensorflow.keras.applications import MobileNetV2, VGG16, DenseNet121
from tensorflow.keras.layers import Dense, GlobalAveragePooling2D, Dropout
from tensorflow.keras.models import Model
from sklearn.metrics import classification_report, precision_recall_fscore_support

def build_model(arch):
    if arch == 'MobileNetV2':
        base = MobileNetV2(weights='imagenet', include_top=False, input_shape=(224, 224, 3))
    elif arch == 'VGG16':
        base = VGG16(weights='imagenet', include_top=False, input_shape=(224, 224, 3))
    elif arch == 'DenseNet121':
        base = DenseNet121(weights='imagenet', include_top=False, input_shape=(224, 224, 3))
    
    for layer in base.layers:
        layer.trainable = False
        
    x = GlobalAveragePooling2D()(base.output)
    x = Dense(256, activation='relu')(x)
    x = Dropout(0.3)(x)
    out = Dense(10, activation='softmax')(x)
    
    model = Model(inputs=base.input, outputs=out)
    model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])
    return model

models = {}
histories = {}
results = {}

for arch in ['MobileNetV2', 'VGG16', 'DenseNet121']:
    print(f'Training {arch}...')
    m = build_model(arch)
    h = m.fit(train_gen, validation_data=val_gen, epochs=3, verbose=1)
    models[arch] = m
    histories[arch] = h
    
    preds = m.predict(val_gen)
    y_pred = np.argmax(preds, axis=1)
    y_true = val_gen.classes
    p, r, f1, _ = precision_recall_fscore_support(y_true, y_pred, average='weighted', zero_division=0)
    acc = np.mean(y_pred == y_true)
    results[arch] = {'Accuracy': acc, 'Precision': p, 'Recall': r, 'F1-Score': f1}

os.makedirs('models', exist_ok=True)
best_arch = max(results, key=lambda k: results[k]['F1-Score'])
models[best_arch].save('models/model.h5')
print(f'[SUCCESS] Saved Best Model ({best_arch}) to models/model.h5')
```

## Step 5: Model Evaluation & Comparison

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.DataFrame(results).T
print('=== MODEL PERFORMANCE COMPARISON ===')
    print(df)

plt.figure(figsize=(10, 4))
for arch in histories:
    plt.plot(histories[arch].history['val_accuracy'], label=f'{arch} Val Acc')
plt.title('Validation Accuracy Comparison')
plt.xlabel('Epochs')
plt.ylabel('Accuracy')
plt.legend()
plt.show()
```
